```python
                                                                     #CATEGORY 8


#1. Medical Diagnosis




# Medical Diagnosis using Knowledge Base

facts = {"fever", "cough", "body_pain"}

rules = {
    "Flu": {"fever", "cough", "body_pain"},
    "Cold": {"cough", "sneezing"},
    "Malaria": {"fever", "chills", "sweating"}
}

# Forward Chaining
print("Forward Chaining:")
for disease, symptoms in rules.items():
    if symptoms.issubset(facts):
        print("Possible diagnosis:", disease)

# Backward Chaining
def backward_chaining(goal):
    if goal in rules:
        if rules[goal].issubset(facts):
            return True
    return False

print("\nBackward Chaining:")
goal = "Flu"

if backward_chaining(goal):
    print(goal, "is supported by the symptoms")
else:
    print(goal, "is not supported by the symptoms")
```

```python
#2. Autonomous Vehicle




# Autonomous Vehicle using Knowledge Base

facts = {"red_light", "pedestrian"}

rules = {
    "STOP": {"red_light"},
    "SLOW_DOWN": {"pedestrian"},
    "MOVE": {"green_light", "road_clear"}
}

# Forward Chaining
print("Forward Chaining:")
for action, conditions in rules.items():
    if conditions.issubset(facts):
        print("Vehicle action:", action)

# Backward Chaining
def backward_chaining(goal):
    if goal in rules:
        return rules[goal].issubset(facts)
    return False

print("\nBackward Chaining:")
goal = "STOP"

if backward_chaining(goal):
    print("Goal achieved:", goal)
else:
    print("Goal not achieved:", goal)
```
