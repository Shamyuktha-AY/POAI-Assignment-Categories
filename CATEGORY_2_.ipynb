```python
#category 2

#Fully Observable Environment

#Treasure Hunt Robot


treasure = 4
obstacles = [2]

position = int(input("Robot Position (1-5): "))

while position != treasure:
    if position + 1 in obstacles:
        print("Obstacle Ahead! Jump")
        position += 2
    else:
        print("Move Forward")
        position += 1

print("Treasure Found!")
```

```python
#Static Environment

#Project: Smart Hospital Robot


class HospitalRobot:
    def __init__(self):
        self.rooms = {
            "101": "Patient",
            "102": "Empty",
            "103": "Patient",
            "104": "Medicine Store",
            "105": "Doctor"
        }

    def display_rooms(self):
        print("\nHospital Map")
        for room, status in self.rooms.items():
            print("Room", room, ":", status)

    def deliver_medicine(self, room):
        if room in self.rooms:
            if self.rooms[room] == "Patient":
                print("Medicine Delivered to Room", room)
            else:
                print("No Patient in Room", room)
        else:
            print("Invalid Room Number")

robot = HospitalRobot()

robot.display_rooms()

room = input("\nEnter Room Number: ")
robot.deliver_medicine(room)
```

```python
#Dynamic environment


class HospitalRobot:
    def __init__(self):
        self.rooms = {
            "101": "Patient",
            "102": "Empty",
            "103": "Patient",
            "104": "Medicine Store",
            "105": "Doctor"
        }

    def display(self):
        print("\nCurrent Hospital Status")
        for room, status in self.rooms.items():
            print(room, ":", status)

    def update_room(self, room, status):
        self.rooms[room] = status

    def deliver(self, room):
        if self.rooms.get(room) == "Patient":
            print("Medicine Delivered")
        else:
            print("Patient Not Found")

robot = HospitalRobot()

while True:
    robot.display()

    print("\n1. Deliver Medicine")
    print("2. Update Room")
    print("3. Exit")

    choice = int(input("Enter Choice: "))

    if choice == 1:
        room = input("Enter Room: ")
        robot.deliver(room)

    elif choice == 2:
        room = input("Enter Room: ")
        status = input("Enter New Status (Patient/Empty): ")
        robot.update_room(room, status)

    elif choice == 3:
        break
```

```python
#Discrete Environment


#Elevator System


class Elevator:
    def __init__(self):
        self.current_floor = 1

    def move(self, destination):
        while self.current_floor != destination:
            if self.current_floor < destination:
                self.current_floor += 1
            else:
                self.current_floor -= 1

            print("Current Floor:", self.current_floor)

        print("Destination Reached!")

elevator = Elevator()

destination = int(input("Enter destination floor (1-5): "))

if 1 <= destination <= 5:
    elevator.move(destination)
else:
    print("Invalid Floor")
```

```python
#Continuous Environment


#Cruise Control System


class CruiseControl:
    def __init__(self):
        self.speed = 0.0

    def set_speed(self, target):
        while abs(self.speed - target) > 0.5:
            if self.speed < target:
                self.speed += 2.5
            else:
                self.speed -= 2.5

            print("Current Speed: {:.1f} km/h".format(self.speed))

        print("Target Speed Reached")

car = CruiseControl()

target = float(input("Enter target speed: "))
car.set_speed(target)
```

```python
#Deterministic Environment


#Robot Navigation


class Robot:
    def __init__(self):
        self.position = 0

    def move(self, direction):
        if direction == "right":
            self.position += 1
        elif direction == "left":
            self.position -= 1

        print("Current Position:", self.position)

robot = Robot()

while True:
    print("\n1. Move Right")
    print("2. Move Left")
    print("3. Exit")

    choice = int(input("Enter Choice: "))

    if choice == 1:
        robot.move("right")
    elif choice == 2:
        robot.move("left")
    elif choice == 3:
        print("Program Ended")
        break
    else:
        print("Invalid Choice")
```

```python
#Stochastic Environment


#Robot Navigation with Random Obstacles



import random

class Robot:
    def __init__(self):
        self.position = 0

    def move(self):
        chance = random.randint(1, 2)

        if chance == 1:
            self.position += 1
            print("Moved Successfully")
        else:
            print("Obstacle Found! Stayed in Same Position")

        print("Current Position:", self.position)

robot = Robot()

while True:
    print("\n1. Move")
    print("2. Exit")

    choice = int(input("Enter Choice: "))

    if choice == 1:
        robot.move()
    elif choice == 2:
        print("Program Ended")
        break
```

```python
#Episodic Environment


#Email Spam Classifier



class SpamFilter:

    def classify(self, email):
        if "win" in email.lower() or "prize" in email.lower():
            return "Spam"
        else:
            return "Not Spam"

agent = SpamFilter()

while True:
    email = input("\nEnter email text (or 'exit'): ")

    if email.lower() == "exit":
        break

    print("Result:", agent.classify(email))
```
