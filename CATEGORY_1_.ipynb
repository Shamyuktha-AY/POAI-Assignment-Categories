```python
                #SIMPLE REFLEX AGENT
#VACCUM CLEANER

#A Simple Reflex Agent acts only on the current percept (input). It uses if-else rules and does not remember previous states.

class SimpleReflexAgent:

    def act(self, location, status):
        if status == "Dirty":
            return "Suck"
        elif location == "A":
            return "Move Right"
        elif location == "B":
            return "Move Left"
        else:
            return "No Action"



agent = SimpleReflexAgent()


location = input("Enter location (A/B): ")
status = input("Enter status (Dirty/Clean): ")


action = agent.act(location, status)

print("Agent Action:", action)
```

```python
#Traffic Light Agent

signal = input("Enter signal (Red/Yellow/Green): ")

if signal == "Red":
    print("Action: Stop")
elif signal == "Yellow":
    print("Action: Slow Down")
elif signal == "Green":
    print("Action: Go")
else:
    print("Invalid Signal")
```

```python
\                 #MODEL BASED AGENT

#A Model-Based Agent remembers the previous state (internal model) and uses both the current percept and stored information to make decisions.

#SMART AC

class SmartAC:
    def __init__(self):
        self.ac = "OFF"

    def control(self, temp):
        if temp > 30:
            self.ac = "ON"
        elif temp < 25:
            self.ac = "OFF"

        print("AC Status:", self.ac)

ac = SmartAC()

while True:
    t = int(input("Temperature: "))
    ac.control(t)
```

```python
#WATER TANK MONITORING


class WaterTank:
    def __init__(self):
        self.motor = "OFF"

    def check(self, level):
        if level == "Low":
            self.motor = "ON"
        elif level == "High":
            self.motor = "OFF"

        print("Motor Status:", self.motor)

tank = WaterTank()

while True:
    level = input("Water Level (Low/High): ")
    tank.check(level)
```

```python
           #GOAL BASED AGENT
#A Goal-Based Agent chooses actions that help it achieve a specific goal. It compares the current state with the goal and decides what action to take.

#TEMPERATURE GOAL

goal = 25
temp = int(input("Enter current temperature: "))

while temp != goal:
    if temp > goal:
        print("Turn ON AC")
        temp -= 1
    else:
        print("Turn ON Heater")
        temp += 1

print("Desired Temperature Reached")
```

```python
#ELEVATOR GOAL



goal = int(input("Enter destination floor: "))
current = int(input("Enter current floor: "))

while current != goal:
    if current < goal:
        print("Moving Up")
        current += 1
    else:
        print("Moving Down")
        current -= 1

print("Destination Reached!")
```

```python
              #UTILITY BASED AGENT


#A Utility-Based Agent chooses the action that gives the highest utility (benefit or score) when there are multiple possible actions. Instead of just reaching a goal, it tries to achieve the best outcome.

#SMART FAN AGENT

low = 50
medium = 80
high = 70

if medium >= low and medium >= high:
    print("Choose Medium Speed")
elif high >= low:
    print("Choose High Speed")
else:
    print("Choose Low Speed")
```

```python
#ROUTE SELECTION AGENT

routes = {
    "Route A": 70,
    "Route B": 90,
    "Route C": 80
}

best_route = max(routes, key=routes.get)

print("Available Routes:")
for route, utility in routes.items():
    print(route, "Utility:", utility)

print("Best Route:", best_route)
```
