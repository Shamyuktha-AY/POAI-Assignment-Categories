```python
                                                                #CATEGORY 6


#1. Min-Max Search Algorithm for Chess Game



# Min-Max Algorithm for Chess

def minimax(depth, node, is_max):
    if depth == 0:
        return node

    if is_max:
        return max(
            minimax(depth - 1, child, False)
            for child in node
        )
    else:
        return min(
            minimax(depth - 1, child, True)
            for child in node
        )

# Example game tree with evaluation scores
tree = [
    [3, 5],
    [2, 9]
]

best_score = minimax(2, tree, True)

print("Best score for MAX:", best_score)
```

```python
#2. Alpha-Beta Pruning Algorithm for Chess Game



# Alpha-Beta Pruning for Chess

def alphabeta(depth, node, alpha, beta, is_max):
    if depth == 0:
        return node

    if is_max:
        best = float('-inf')

        for child in node:
            value = alphabeta(
                depth - 1, child, alpha, beta, False
            )
            best = max(best, value)
            alpha = max(alpha, best)

            if beta <= alpha:
                break

        return best

    else:
        best = float('inf')

        for child in node:
            value = alphabeta(
                depth - 1, child, alpha, beta, True
            )
            best = min(best, value)
            beta = min(beta, best)

            if beta <= alpha:
                break

        return best

tree = [
    [3, 5],
    [2, 9]
]

best_score = alphabeta(
    2, tree, float('-inf'), float('inf'), True
)

print("Best score for MAX:", best_score)
```

```python
#3. Min-Max Search Algorithm for Tic-Tac-Toe



# Min-Max Algorithm for Tic-Tac-Toe

board = ["X", "O", "X",
         "O", "X", " ",
         " ", " ", "O"]

def winner(b):
    lines = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]

    for a, c, d in lines:
        if b[a] == b[c] == b[d] != " ":
            return b[a]

    return None

def minimax(b, is_max):
    result = winner(b)

    if result == "X":
        return 1
    if result == "O":
        return -1
    if " " not in b:
        return 0

    if is_max:
        best = -float('inf')

        for i in range(9):
            if b[i] == " ":
                b[i] = "X"
                best = max(best, minimax(b, False))
                b[i] = " "

        return best

    else:
        best = float('inf')

        for i in range(9):
            if b[i] == " ":
                b[i] = "O"
                best = min(best, minimax(b, True))
                b[i] = " "

        return best

best_score = -float('inf')
best_move = -1

for i in range(9):
    if board[i] == " ":
        board[i] = "X"
        score = minimax(board, False)
        board[i] = " "

        if score > best_score:
            best_score = score
            best_move = i

print("Best move for X:", best_move)
board[best_move] = "X"

for i in range(0, 9, 3):
    print("|".join(board[i:i+3]))
```
