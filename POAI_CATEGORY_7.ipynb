```python
                                                         #CATEGORY 7


#1. Map Coloring


# Map Coloring using CSP

colors = ["Red", "Green", "Blue"]

neighbors = {
    "A": ["B", "C"],
    "B": ["A", "C", "D"],
    "C": ["A", "B", "D"],
    "D": ["B", "C"]
}

assignment = {}

def is_safe(region, color):
    for neighbor in neighbors[region]:
        if assignment.get(neighbor) == color:
            return False
    return True

def solve(regions):
    if not regions:
        return True

    region = regions[0]

    for color in colors:
        if is_safe(region, color):
            assignment[region] = color

            if solve(regions[1:]):
                return True

            del assignment[region]

    return False

if solve(list(neighbors.keys())):
    print("Map Coloring Solution:")
    for region, color in assignment.items():
        print(region, ":", color)
else:
    print("No solution")
```

```python
#2. Crypt-Arithmetic Puzzle



# Crypt-Arithmetic: SEND + MORE = MONEY

from itertools import permutations

letters = "SENDMORY"

for digits in permutations(range(10), len(letters)):
    d = dict(zip(letters, digits))

    if d["S"] == 0 or d["M"] == 0:
        continue

    SEND = 1000*d["S"] + 100*d["E"] + 10*d["N"] + d["D"]
    MORE = 1000*d["M"] + 100*d["O"] + 10*d["R"] + d["E"]
    MONEY = 10000*d["M"] + 1000*d["O"] + 100*d["N"] + 10*d["E"] + d["Y"]

    if SEND + MORE == MONEY:
        print("SEND =", SEND)
        print("MORE =", MORE)
        print("MONEY =", MONEY)
        break
```

```python
#3. Local Search for CSP


# Local Search for CSP using Min-Conflicts

import random

colors = ["Red", "Green", "Blue"]

neighbors = {
    "A": ["B", "C"],
    "B": ["A", "C", "D"],
    "C": ["A", "B", "D"],
    "D": ["B", "C"]
}

assignment = {
    region: random.choice(colors)
    for region in neighbors
}

def conflicts(region, color):
    return sum(
        assignment.get(n) == color
        for n in neighbors[region]
    )

for step in range(100):
    conflicted = [
        r for r in neighbors
        if conflicts(r, assignment[r]) > 0
    ]

    if not conflicted:
        break

    region = random.choice(conflicted)

    best_color = min(
        colors,
        key=lambda c: conflicts(region, c)
    )

    assignment[region] = best_color

print("Final Assignment:")
for region, color in assignment.items():
    print(region, ":", color)

if all(conflicts(r, assignment[r]) == 0 for r in neighbors):
    print("Solution Found")
else:
    print("Try Again")
```
