```python
#1. Breadth First Search (BFS)


#Use Case: Finding the shortest path in a social network



from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

def bfs(graph, start, goal):
    queue = deque([[start]])
    visited = []

    while queue:
        path = queue.popleft()
        node = path[-1]

        if node not in visited:
            visited.append(node)

            if node == goal:
                return path

            for neighbor in graph[node]:
                new_path = list(path)
                new_path.append(neighbor)
                queue.append(new_path)

    return None

start = 'A'
goal = 'F'

result = bfs(graph, start, goal)
print("BFS Path:", result)
```

```python
#2. Depth First Search (DFS)


#Use Case:

#Explore all possible paths deeply before backtracking.


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

def dfs(graph, node, goal, visited=None):
    if visited is None:
        visited = []

    visited.append(node)

    if node == goal:
        return visited

    for neighbor in graph[node]:
        if neighbor not in visited:
            result = dfs(graph, neighbor, goal, visited)
            if result:
                return result

    return None

print("DFS Path:", dfs(graph, 'A', 'F'))
```

```python
#3. Depth Limited Search (DLS)


#Use Case:

#Search only up to a fixed depth.


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

def dls(node, goal, limit, path=[]):
    path = path + [node]

    if node == goal:
        return path

    if limit <= 0:
        return None

    for neighbor in graph[node]:
        result = dls(neighbor, goal, limit - 1, path)
        if result:
            return result

    return None

depth_limit = 2
result = dls('A', 'F', depth_limit)

print("DLS Result:", result)
```

```python
#4. Bidirectional Search


#Use Case:

#Search simultaneously from start and goal.


from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

def bidirectional_search(start, goal):
    if start == goal:
        return [start]

    front = {start}
    back = {goal}

    visited_front = {start: None}
    visited_back = {goal: None}

    while front and back:
        next_front = set()

        for node in front:
            for neighbor in graph[node]:
                if neighbor not in visited_front:
                    visited_front[neighbor] = node
                    next_front.add(neighbor)

                    if neighbor in visited_back:
                        return neighbor

        front = next_front

    return None

meeting = bidirectional_search('A', 'F')

if meeting:
    print("Meeting Node:", meeting)
else:
    print("No Path Found")
```

```python


## Comparison Table

| Algorithm     | Data Structure  | Complete | Optimal | Time Complexity |
| ------------- | --------------- | -------- | ------- | --------------- |
| BFS           | Queue           | Yes      | Yes     | O(V + E)        |
| DFS           | Stack/Recursion | No       | No      | O(V + E)        |
| DLS           | Stack + Limit   | No       | No      | O(b^l)          |
| Bidirectional | Two Queues      | Yes      | Yes     | O(b^(d/2))      |

Where:

* **V** = Vertices
* **E** = Edges
* **b** = Branching Factor
* **d** = Depth
* **l** = Depth Limit

These examples are simple and suitable for your **AI Lab Record and Viva**.
```
